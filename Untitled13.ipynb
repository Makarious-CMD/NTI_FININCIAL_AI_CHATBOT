```python

import nltk
from nltk.tokenize import sent_tokenize
from transformers import pipeline
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM


nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

summarizer_name = "Sachin21112004/distilbart-news-summarizer"
tokenizer = AutoTokenizer.from_pretrained(summarizer_name)
summarizer_model = AutoModelForSeq2SeqLM.from_pretrained(summarizer_name)

ner_model = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
sentiment_model = pipeline("sentiment-analysis", model="mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis")


def analyze_any_news(text):
    input_word_count = len(text.split())
    dynamic_max = max(30, int(input_word_count * 0.4))
    dynamic_min = max(10, int(input_word_count * 0.1))


    inputs = tokenizer(text, return_tensors="pt", max_length=1024, truncation=True)


    summary_ids = summarizer_model.generate(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_length=dynamic_max,
        min_length=dynamic_min,
        do_sample=False
    )

    summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

    sentences = sent_tokenize(summary_text)
    bullet_points = "\n".join([f"- {s.strip()}" for s in sentences if s.strip()])


    entities_raw = ner_model(text)
    found_entities = {}

    for ent in entities_raw:
        label = ent['entity_group']
        word = ent['word']

        if label not in found_entities:
            found_entities[label] = set()

        found_entities[label].add(word)


    sentiment = sentiment_model(text)[0]
    sentiment_map = {"positive": "positive", "negative": "negative", "natural": "natural"}
    sentiment_label = sentiment_map.get(sentiment['label'].lower(), sentiment['label'])



    print(f"{sentiment_label} ({round(sentiment['score']*100, 1)}%)\n")


    print(bullet_points + "\n")


    if not found_entities:
        print("")
    else:
        for label, words in found_entities.items():
            print(f" {label}: {', '.join(words)}")

    print("="*50 + "\n")



any_news_you_want = """
Kuwait is creating specialist court divisions to handle commercial and financial disputes in an effort to improve its appeal to foreign investors by speeding up contract enforcement
"""

analyze_any_news(any_news_you_want)
```
